```python
# ============================================================
# CREDIT CARD FRAUD DETECTION
# PHASE 1 — DATA AUDIT
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path


# ============================================================
# 1. LOAD DATA
# ============================================================

DATA_PATH = Path("../data/creditcard.csv")

df = pd.read_csv(DATA_PATH)

print("=" * 70)
print("DATASET LOADED")
print("=" * 70)

print(f"Dataset shape       : {df.shape}")
print(f"Number of records   : {len(df):,}")
print(f"Number of columns   : {len(df.columns)}")
print(f"Number of features  : {len(df.columns) - 1}")


# ============================================================
# 2. FIRST LOOK
# ============================================================

print("\n" + "=" * 70)
print("FIRST 5 TRANSACTIONS")
print("=" * 70)

display(df.head())


# ============================================================
# 3. COLUMN INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("COLUMN INFORMATION")
print("=" * 70)

print(df.dtypes)


# ============================================================
# 4. MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

missing = df.isna().sum()

print(f"Total missing values: {missing.sum():,}")

if missing.sum() == 0:
    print("✓ No missing values found.")
else:
    print("\nColumns containing missing values:")
    print(missing[missing > 0])


# ============================================================
# 5. DUPLICATES
# ============================================================

print("\n" + "=" * 70)
print("DUPLICATE RECORDS")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print(f"Duplicate rows: {duplicate_count:,}")


# ============================================================
# 6. TARGET / FRAUD DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("FRAUD DISTRIBUTION")
print("=" * 70)

class_counts = df["Class"].value_counts().sort_index()

class_percentages = (
    df["Class"]
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

fraud_summary = pd.DataFrame({
    "Count": class_counts,
    "Percentage": class_percentages
})

fraud_summary.index = ["Legitimate", "Fraud"]

display(fraud_summary)

fraud_rate = df["Class"].mean()

print(f"\nFraud rate: {fraud_rate:.4%}")


# ============================================================
# 7. BASIC STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

display(df.describe().T)


# ============================================================
# 8. TIME STRUCTURE
# ============================================================

print("\n" + "=" * 70)
print("TIME STRUCTURE")
print("=" * 70)

print(f"Minimum Time: {df['Time'].min():,.0f} seconds")
print(f"Maximum Time: {df['Time'].max():,.0f} seconds")

observation_hours = df["Time"].max() / 3600

print(f"Observation window: {observation_hours:.2f} hours")


# Convert Time from seconds to hours
df["Time_hours"] = df["Time"] / 3600


# ============================================================
# 9. FRAUD OVER TIME
# ============================================================

plt.figure(figsize=(12, 5))

plt.scatter(
    df.loc[df["Class"] == 0, "Time_hours"],
    np.zeros((df["Class"] == 0).sum()),
    s=2,
    alpha=0.10,
    label="Legitimate"
)

plt.scatter(
    df.loc[df["Class"] == 1, "Time_hours"],
    np.ones((df["Class"] == 1).sum()),
    s=10,
    alpha=0.70,
    label="Fraud"
)

plt.xlabel("Time since first transaction (hours)")
plt.ylabel("Transaction class")

plt.yticks(
    [0, 1],
    ["Legitimate", "Fraud"]
)

plt.title("Fraud Transactions Across Observation Window")

plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# 10. TRANSACTION AMOUNT
# ============================================================

print("\n" + "=" * 70)
print("TRANSACTION AMOUNT")
print("=" * 70)

amount_summary = df.groupby("Class")["Amount"].describe()

display(amount_summary)


# ============================================================
# 11. TRANSACTION AMOUNT DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 5))

plt.hist(
    df.loc[df["Class"] == 0, "Amount"],
    bins=100,
    alpha=0.6,
    label="Legitimate"
)

plt.hist(
    df.loc[df["Class"] == 1, "Amount"],
    bins=100,
    alpha=0.8,
    label="Fraud"
)

plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.title("Transaction Amount Distribution")

plt.legend()
plt.tight_layout()

plt.show()


# ============================================================
# 12. LOG-TRANSFORMED AMOUNT
# ============================================================

plt.figure(figsize=(10, 5))

plt.hist(
    np.log1p(df["Amount"]),
    bins=100
)

plt.xlabel("log(1 + Transaction Amount)")
plt.ylabel("Frequency")

plt.title("Log-Transformed Transaction Amount")

plt.tight_layout()

plt.show()


# ============================================================
# 13. FRAUD RATE BY TRANSACTION AMOUNT
# ============================================================

df["Amount_bin"] = pd.qcut(
    df["Amount"],
    q=10,
    duplicates="drop"
)

amount_fraud_rate = (
    df.groupby("Amount_bin", observed=True)["Class"]
    .mean()
    * 100
)

print("\n" + "=" * 70)
print("FRAUD RATE BY TRANSACTION AMOUNT DECILE")
print("=" * 70)

display(amount_fraud_rate.to_frame("Fraud Rate (%)"))


# ============================================================
# 14. FRAUD BY TIME PERIOD
# ============================================================

df["Time_bin"] = pd.qcut(
    df["Time"],
    q=10,
    duplicates="drop"
)

time_fraud_rate = (
    df.groupby("Time_bin", observed=True)["Class"]
    .agg(
        fraud_rate="mean",
        transactions="count",
        fraud_count="sum"
    )
)

time_fraud_rate["fraud_rate"] *= 100

print("\n" + "=" * 70)
print("FRAUD RATE BY TIME DECILE")
print("=" * 70)

display(time_fraud_rate)


# ============================================================
# 15. FINAL AUDIT SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL DATA AUDIT SUMMARY")
print("=" * 70)

print(f"Transactions              : {len(df):,}")
print(f"Features                  : {len(df.columns) - 3}")
print(f"Fraud transactions        : {(df['Class'] == 1).sum():,}")
print(f"Legitimate transactions   : {(df['Class'] == 0).sum():,}")
print(f"Fraud rate                : {df['Class'].mean():.4%}")
print(f"Missing values            : {df.isna().sum().sum():,}")
print(f"Duplicate rows            : {df.duplicated().sum():,}")
print(f"Observation window        : {observation_hours:.2f} hours")

print("\n✓ PHASE 1 DATA AUDIT COMPLETE")
```
